```python
import pandas as pd
import time
import matplotlib.pyplot as plt
from collections import Counter

data_path = '../../02_ml_basics/data/ratings_train.txt'  # Adjust as needed
try:
    df = pd.read_csv(data_path, sep='\t', encoding='utf-8')
except Exception:
    df = pd.read_csv(data_path, sep=',', encoding='utf-8')
print('Shape:', df.shape)
df.head()
```

```python
def simulate_streaming(df, delay=0.2, max_events=20):
    label_counts = Counter()
    plt.ion()
    fig, ax = plt.subplots()
    for idx, row in df.iterrows():
        label = row.get('label', None) or row.get('sentiment', None)
        if label is not None:
            label_counts[label] += 1
        ax.clear()
        ax.bar(label_counts.keys(), label_counts.values())
        ax.set_title('Running Label Counts')
        plt.draw()
        plt.pause(0.01)
        print(f'[EVENT {idx}] {row.to_dict()}')
        time.sleep(delay)
        if idx + 1 >= max_events:
            break
    plt.ioff()
    plt.show()

simulate_streaming(df, delay=0.2, max_events=20)
```
