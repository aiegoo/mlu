````python
# 🎉 VERIFY GPU ACCELERATION IS NOW WORKING
print("=== 🚀 GPU ACCELERATION VERIFICATION ===")

import torch
import time
import platform
import subprocess

print(f"🔥 PyTorch version: {torch.__version__}")
print(f"🖥️ Platform: {platform.platform()}")

# Check CUDA availability
if torch.cuda.is_available():
    print("✅ CUDA is available!")
    device_count = torch.cuda.device_count()
    print(f"🎮 Number of GPUs: {device_count}")
    
    for i in range(device_count):
        gpu_name = torch.cuda.get_device_name(i)
        gpu_properties = torch.cuda.get_device_properties(i)
        gpu_memory = gpu_properties.total_memory / 1024**3
        print(f"  GPU {i}: {gpu_name}")
        print(f"    💾 Memory: {gpu_memory:.1f} GB")
        print(f"    🧮 Compute Capability: {gpu_properties.major}.{gpu_properties.minor}")
    
    # Set default device to GPU
    device = torch.device('cuda')
    print(f"🎯 Default device set to: {device}")
    
    # Performance comparison test
    print("\n=== ⚡ PERFORMANCE COMPARISON ===")
    
    # Test matrix size
    size = 2000
    iterations = 5
    
    # CPU test
    print("🖥️ CPU Performance Test:")
    x_cpu = torch.randn(size, size)
    y_cpu = torch.randn(size, size)
    
    start_time = time.time()
    for _ in range(iterations):
        result_cpu = torch.matmul(x_cpu, y_cpu)
    cpu_time = (time.time() - start_time) / iterations
    print(f"   Average time: {cpu_time:.4f} seconds")
    
    # GPU test
    print("🎮 GPU Performance Test:")
    x_gpu = torch.randn(size, size, device=device)
    y_gpu = torch.randn(size, size, device=device)
    
    # Warm up
    torch.matmul(x_gpu, y_gpu)
    torch.cuda.synchronize()
    
    start_time = time.time()
    for _ in range(iterations):
        result_gpu = torch.matmul(x_gpu, y_gpu)
    torch.cuda.synchronize()
    gpu_time = (time.time() - start_time) / iterations
    print(f"   Average time: {gpu_time:.4f} seconds")
    
    # Calculate speedup
    speedup = cpu_time / gpu_time
    print(f"\n🚀 GPU Speedup: {speedup:.2f}x faster!")
    
    # Memory info
    print(f"\n📊 GPU Memory Usage:")
    print(f"   Allocated: {torch.cuda.memory_allocated() / 1024**2:.1f} MB")
    print(f"   Cached: {torch.cuda.memory_reserved() / 1024**2:.1f} MB")
    
    print("\n🎉 RTX 4050 GPU acceleration is working perfectly!")
    print("💡 All D2L.ai operations will now use GPU acceleration")
    
else:
    print("❌ CUDA is still not available")
    print("🔧 Please check NVIDIA drivers and CUDA installation")

print(f"\n=== 🎯 RECOMMENDED DEVICE FOR D2L.AI ===")
if torch.cuda.is_available():
    print("✅ Use: device = torch.device('cuda')")
    print("🎮 GPU acceleration will significantly speed up training!")
else:
    print("⚠️ Use: device = torch.device('cpu')")
    print("🐌 Consider Docker container for GPU access")
````

### ⚡ Tensor Operations - The Core Skills

````python
# Arithmetic operations - elementwise by default
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])

print("x:", x)
print("y:", y)
print("x + y:", x + y)
print("x - y:", x - y) 
print("x * y:", x * y)
print("x / y:", x / y)
print("x ** y:", x ** y)  # Exponentiation
````

### 🔗 Concatenation and Splitting

````python
# Concatenation - joining tensors
X = torch.arange(12, dtype=torch.float32).reshape((3, 4))
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])

print("X:")
print(X)
print("\nY:")
print(Y)

# Concatenate along different dimensions
print("\nConcatenate along rows (dim=0):")
print(torch.cat((X, Y), dim=0))

print("\nConcatenate along columns (dim=1):")
print(torch.cat((X, Y), dim=1))
````

### 🎯 Indexing and Slicing - Accessing Data

````python
# Indexing and slicing - crucial for data manipulation
X = torch.arange(12, dtype=torch.float32).reshape((3, 4))
print("Original tensor:")
print(X)

# Access elements
print("\nElement at [1, 2]:", X[1, 2])
print("Last row:", X[-1])
print("Rows 1-2, columns 1-3:")
print(X[1:3, 1:3])

# Modify elements
X[1, 2] = 17
print("\nAfter modifying X[1, 2] = 17:")
print(X)

# Modify multiple elements
X[:2, :] = 12  # First two rows
print("\nAfter setting first two rows to 12:")
print(X)
````

### 🔄 Memory Efficiency - Important for Large Models

````python
# Understanding memory allocation
before = id(Y)
Y = Y + X
print("Memory address changed:", id(Y) != before)

# In-place operations to save memory
Z = torch.zeros_like(Y)
print("\nBefore in-place operation, id(Z):", id(Z))
Z[:] = X + Y  # In-place assignment
print("After in-place operation, id(Z):", id(Z))

# Alternative in-place operations
X += Y  # Equivalent to X[:] = X + Y
print("\nIn-place addition result:")
print(X)
````

### 🔄 Converting to Other Formats

## 🧮 Chapter 2.3: Linear Algebra - Mathematical Foundations

**Essential math for understanding deep learning**

### 🔗 Official D2L Section: [2.3 Linear Algebra](https://d2l.ai/chapter_preliminaries/linear-algebra.html)

### 📐 Scalars, Vectors, Matrices, and Tensors

````python
# Scalars - 0-dimensional tensors
scalar = torch.tensor(3.0)
print(f"Scalar: {scalar}, shape: {scalar.shape}, ndim: {scalar.ndim}")

# Vectors - 1-dimensional tensors
vector = torch.tensor([1, 2, 3, 4])
print(f"\nVector: {vector}")
print(f"Shape: {vector.shape}, length: {len(vector)}")

# Matrices - 2-dimensional tensors
matrix = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(f"\nMatrix:\n{matrix}")
print(f"Shape: {matrix.shape}")

# Tensors - n-dimensional arrays
tensor_3d = torch.arange(24).reshape(2, 3, 4)
print(f"\n3D Tensor shape: {tensor_3d.shape}")
print(f"Number of dimensions: {tensor_3d.ndim}")
````

### ➕ Basic Operations

````python
# Matrix transpose
A = torch.tensor([[1, 2, 3], [4, 5, 6]])
print("Original matrix A:")
print(A)
print("\nTranspose A.T:")
print(A.T)

# Symmetric matrices
symmetric = torch.tensor([[1, 2, 3], [2, 0, 4], [3, 4, 5]])
print("\nSymmetric matrix:")
print(symmetric)
print("\nIs symmetric?", torch.equal(symmetric, symmetric.T))
````

### 📊 Reduction Operations - Summarizing Data

````python
# Sum operations
x = torch.arange(3, dtype=torch.float32)
print(f"Vector x: {x}")
print(f"Sum of x: {x.sum()}")

# Matrix sum operations
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
print(f"\nMatrix A:\n{A}")
print(f"Sum of all elements: {A.sum()}")
print(f"Sum along axis 0 (columns): {A.sum(axis=0)}")
print(f"Sum along axis 1 (rows): {A.sum(axis=1)}")

# Other reduction operations
print(f"\nMean: {A.mean()}")
print(f"Mean along axis 0: {A.mean(axis=0)}")
print(f"Mean along axis 1: {A.mean(axis=1)}")
````

### 🔢 Dot Products - Fundamental Operation

````python
# Vector dot product
x = torch.tensor([1, 2, 3], dtype=torch.float32)
y = torch.tensor([4, 5, 6], dtype=torch.float32)

dot_product = torch.dot(x, y)
print(f"x: {x}")
print(f"y: {y}")
print(f"Dot product x·y: {dot_product}")
print(f"Manual calculation: {(x * y).sum()}")

# Geometric interpretation
x_norm = torch.norm(x)
y_norm = torch.norm(y)
print(f"\n||x||: {x_norm}")
print(f"||y||: {y_norm}")
print(f"Cosine similarity: {dot_product / (x_norm * y_norm)}")
````

### 🔄 Matrix-Vector Products

````python
# Matrix-vector multiplication (Ax = b)
A = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)
x = torch.tensor([7, 8, 9], dtype=torch.float32)

result = torch.mv(A, x)  # Matrix-vector product
print(f"Matrix A:\n{A}")
print(f"\nVector x: {x}")
print(f"\nMatrix-vector product Ax: {result}")

# Manual calculation for understanding
manual_result = torch.tensor([
    torch.dot(A[0], x),  # First row dot product
    torch.dot(A[1], x)   # Second row dot product
])
print(f"Manual calculation: {manual_result}")
print(f"Results match: {torch.equal(result, manual_result)}")
````

### 🔄 Matrix-Matrix Multiplication

````python
# Matrix multiplication - the heart of deep learning
A = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
B = torch.tensor([[5, 6], [7, 8]], dtype=torch.float32)

# Matrix multiplication
C = torch.mm(A, B)
print(f"Matrix A:\n{A}")
print(f"\nMatrix B:\n{B}")
print(f"\nMatrix multiplication AB:\n{C}")

# Alternative syntax
C_alt = A @ B
print(f"\nUsing @ operator:\n{C_alt}")
print(f"Results match: {torch.equal(C, C_alt)}")

# Show dimensions matter
print(f"\nA shape: {A.shape}")
print(f"B shape: {B.shape}")
print(f"Result shape: {C.shape}")
````

### 📏 Norms - Measuring Magnitude

````python
# Vector norms
x = torch.tensor([3, 4], dtype=torch.float32)
print(f"Vector x: {x}")

# L2 norm (Euclidean norm)
l2_norm = torch.norm(x)
print(f"L2 norm ||x||_2: {l2_norm}")
print(f"Manual L2: {torch.sqrt((x**2).sum())}")

# L1 norm (Manhattan norm)
l1_norm = torch.norm(x, p=1)
print(f"\nL1 norm ||x||_1: {l1_norm}")
print(f"Manual L1: {torch.abs(x).sum()}")

# Frobenius norm for matrices
A = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
frobenius_norm = torch.norm(A)
print(f"\nMatrix A:\n{A}")
print(f"Frobenius norm: {frobenius_norm}")
print(f"Manual Frobenius: {torch.sqrt((A**2).sum())}")
````
